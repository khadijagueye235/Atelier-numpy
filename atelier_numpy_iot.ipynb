## Partie 1 : Listes Python vs tableaux NumPy

**Objectif:**

 Avant d'utiliser NumPy, il faut comprendre *pourquoi* on l'utilise plutôt que les listes Python classiques. Les deux permettent de stocker plusieurs valeurs, mais elles ne se comportent pas du tout pareil face aux opérations mathématiques.

On dispose des températures mesurées par un capteur : `24.5, 25.1, 26.3, 27.0, 25.8`

### 1) Créer une liste Python stockant ces températures

Une **liste Python** est une structure générique : elle peut contenir n'importe quel type de données mélangées (nombres, texte, etc.). On la crée avec des crochets `[ ]`.

```python
temperatures_list = [24.5, 25.1, 26.3, 27.0, 25.8 ]
print(temperatures_list)
```

### 2) Créer un tableau NumPy équivalent

Un **tableau NumPy** (`ndarray`) est optimisé pour les calculs numériques : toutes les valeurs sont du même type, stockées de façon compacte en mémoire, ce qui rend les calculs beaucoup plus rapides. On le crée avec `np.array()`.

```python
temperatures_tab = np.array([24.5, 25.1, 26.3, 27.0, 25.8])
print(temperatures_tab)
```

### 3) Multiplier par 2 les deux collections

* **Liste**

```python
multiplication_list = temperatures_list * 2
print(multiplication_list)
```

* **tableau**

```python
multiplication_tab = temperatures_tab * 2
print(multiplication_tab)
```

### 4) Quelle différence observe-t-on ?


**Explication :**

* Avec une **liste Python**, l'opérateur `* 2` ne multiplie pas les valeurs. Il **répète simplement la liste deux fois**, ce qui fait passer le nombre d'éléments de 5 à 10.
* Avec un **tableau NumPy**, l'opérateur `* 2` effectue une véritable multiplication : **chaque élément du tableau est multiplié par 2**, tout en conservant les 5 éléments.

Cette différence montre l'intérêt des tableaux NumPy pour les calculs numériques. NumPy permet d'effectuer des opérations **élément par élément**, sans avoir besoin d'utiliser une boucle `for`. On parle alors d'**opérations vectorisées**. Cette approche rend le code plus simple et plus adapté à la manipulation de grandes quantités de données, comme les mesures provenant de capteurs.



## Partie 2 : Création des données


**Objectif** 

Avant de pouvoir analyser des données de capteurs, il faut savoir créer des tableaux de différentes façons : tableaux vides à remplir plus tard, tableaux avec une valeur par défaut, tableaux de temps, ou données aléatoires pour simuler des mesures. 
NumPy propose des fonctions toutes prêtes pour chaque cas.


### 1) Créer un tableau représentant les mesures de température de 10 capteurs, initialement à zéro. 

```python
capteurs_zero = np.zeros(10)
print(capteurs_zero)
```

`np.zeros(n)` crée un tableau de `n` valeurs, toutes égales à 0

### 2) Créer un tableau contenant 10 valeurs égales à 1. 

```python
capteurs_un = np.ones(10)
print(capteurs_un)  
```

`np.ones(n)` fonctionne comme `np.zeros`, mais remplit avec des 1 au lieu de des 0.

### 3) Créer un tableau de 10 valeurs initialisées à 25. 

```python
capteurs_25= np.full(10, 25)
print(capteurs_25)
```

`np.full(n, valeur)` crée un tableau de `n` éléments, tous égaux à `valeur`.

### 4) Créer une matrice identité de taille 5 × 5. 

```python
matrice_identite = np.eye(5)
print(matrice_identite)
```

La **matrice identité** est une matrice carrée avec des 1 sur la diagonale et des 0 partout ailleurs. 

 `np.eye(n)` la génère directement.

### 5) Créer les 24 heures (de 0 à 23) de mesure d’une journée 

```python
heures = np.arange(0, 24, 1)
print(heures)
```

`np.arange(debut, fin)` génère une suite de nombres entiers, comme `range()` en Python classique, mais sous forme de tableau NumPy. Attention : la borne de fin (`24`) n'est **pas incluse**, donc on écrit `np.arange(0, 24)` pour obtenir les valeurs de 0 à 23.

### 6) Créer 10 valeurs régulièrement réparties entre 20 °C et 30 °C  

```python
temperatures= np.linspace(20, 30, 10)
print(temperatures)
```

`np.linspace(debut, fin, nombre_de_valeurs)` crée des valeurs **régulièrement espacées** entre deux bornes. Contrairement à `arange`, ici les deux bornes (20 et 30) **sont incluses**, et on précise directement combien de valeurs on veut au total.

### 7) Créer aléatoirement les températures de 100 mesures, en rendant les résultats reproductibles 

Pour générer des nombres aléatoires, on utilise `np.random`. Mais si on relance le code, on obtiendrait normalement des valeurs différentes à chaque fois !

Pour rendre les résultats **reproductibles** (les mêmes valeurs "aléatoires" à chaque exécution), on fixe une **graine aléatoire** avec `np.random.seed(un_nombre)` juste avant de générer les valeurs.

On utilise ensuite `np.random.uniform(min, max, taille)` pour générer des températures réalistes dans une plage donnée.

```python
np.random.seed(42)
mesures_aleatoires = np.random.uniform(20, 60, 100).round(2)
print(mesures_aleatoires)
```

## Partie 3 – Exploration des tableaux 

**Objectif**

 Avant de manipuler un tableau, il est important de savoir "l'interroger" : combien de dimensions a-t-il, combien de valeurs contient-il, de quel type sont ces valeurs, et quelle est sa forme (nombre de lignes/colonnes). Ce sont des réflexes à avoir systématiquement, surtout quand on récupère des données dont on ne connaît pas encore la structure exacte.

### 1) Créer aléatoirement un tableau représentant les 24 températures d'une journée : 

```python
np.random.seed(42)
temp_jour = np.random.uniform(18, 38, 24).round(2)
print(temp_jour)
```

### 2) Déterminer les propriétés du tableau

Chaque tableau NumPy possède des **attributs** qui décrivent sa structure.On y accède sans parenthèses (ce ne sont pas des fonctions à appeler, mais des informations déjà stockées) :

- `.ndim` : nombre de **dimensions** (1 pour une simple liste de valeurs, 2 pour une matrice, etc.)
- `.size` : nombre total de valeurs dans le tableau
- `.dtype` : type des données stockées (ex: `float64` pour des nombres décimaux)
- `.shape` : la forme du tableau, sous forme de tuple (ex: `(24,)` pour 24 valeurs en 1 dimension)

```python
# a) nombre de dimensions
print("Nombre de dimensions :", temp_jour.ndim)
# b) nombre de valeurs
print("Nombre de valeurs :", temp_jour.size)
# c) type des données
print("Type des données :", temp_jour.dtype)
# d) forme du tableau
print("Type des données :", temp_jour.shape)
```

### Partie 4 – Indexation et slicing 

**Objectif**
 Une fois qu'on a un tableau, on a presque toujours besoin d'en extraire une valeur précise ou une portion. En Python (et donc en NumPy), **le premier élément est à l'index 0**, pas 1. On va utiliser le tableau `temp_jour` créé en Partie 3 (les 24 températures d'une journée, aux heures 0 à 23).

### 1) la première mesure :

On accède à un élément avec des crochets `[index]`. Le premier élément est à l'index `0`.

```python
print(temp_jour[0])  
```

### 2) la dernière mesure :

L'index `-1` désigne toujours le **dernier élément**, `-2` l'avant-dernier, etc. Très pratique pour ne pas avoir à compter le nombre exact d'éléments. 

```python
print(temp_jour[-1]) 
 
```

### 3) la température de 8h :

Comme les heures vont de 0 à 23 et que l'index 0 correspond à l'heure 0, **l'heure et l'index coïncident directement** : l'heure 8h est à l'index `8`.

```python
print(temp_jour[8])
```

### 4) les températures de 8h à 12h :

C'est le **slicing** ("découpage") : `tableau[debut:fin]` extrait une partie du tableau. Attention, la borne de fin est **exclue** : pour aller jusqu'à 12h inclus, il faut écrire `13` comme borne de fin.

```python
print(temp_jour[8:13])
```

### 5) les températures des 6 premières heures :

Quand on omet le `debut` dans un slice, Python part automatiquement du tout début du tableau : `[:6]` équivaut à `[0:6]`.

```python
print(temp_jour[:6])
```

### 6) les températures des heures paires

Le slicing accepte un **3e paramètre**, le pas (`step`) : `tableau[debut:fin:pas]`. Ici on veut les heures 0, 2, 4, 6... donc on part de 0 avec un pas de 2. En laissant `fin` vide, ça va jusqu'au bout du tableau.

```python
print(temp_jour[0::2])
```
