## Partie 1 : Listes Python vs tableaux NumPy

**Objectif:**

 Avant d'utiliser NumPy, il faut comprendre *pourquoi* on l'utilise plutôt que les listes Python classiques. Les deux permettent de stocker plusieurs valeurs, mais elles ne se comportent pas du tout pareil face aux opérations mathématiques.

On dispose des températures mesurées par un capteur : `24.5, 25.1, 26.3, 27.0, 25.8`

### 1) Créer une liste Python stockant ces températures

Une **liste Python** est une structure générique : elle peut contenir n'importe quel type de données mélangées (nombres, texte, etc.). On la crée avec des crochets `[ ]`.

```python
temperatures_list = [24.5, 25.1, 26.3, 27.0, 25.8 ]
print(temperatures_list)
```

### 2) Créer un tableau NumPy équivalent

Un **tableau NumPy** (`ndarray`) est optimisé pour les calculs numériques : toutes les valeurs sont du même type, stockées de façon compacte en mémoire, ce qui rend les calculs beaucoup plus rapides. On le crée avec `np.array()`.

```python
temperatures_tab = np.array([24.5, 25.1, 26.3, 27.0, 25.8])
print(temperatures_tab)
```

### 3) Multiplier par 2 les deux collections

* **Liste**

```python
multiplication_list = temperatures_list * 2
print(multiplication_list)
```

* **tableau**

```python
multiplication_tab = temperatures_tab * 2
print(multiplication_tab)
```

### 4) Quelle différence observe-t-on ?


**Explication :**

* Avec une **liste Python**, l'opérateur `* 2` ne multiplie pas les valeurs. Il **répète simplement la liste deux fois**, ce qui fait passer le nombre d'éléments de 5 à 10.
* Avec un **tableau NumPy**, l'opérateur `* 2` effectue une véritable multiplication : **chaque élément du tableau est multiplié par 2**, tout en conservant les 5 éléments.

Cette différence montre l'intérêt des tableaux NumPy pour les calculs numériques. NumPy permet d'effectuer des opérations **élément par élément**, sans avoir besoin d'utiliser une boucle `for`. On parle alors d'**opérations vectorisées**. Cette approche rend le code plus simple et plus adapté à la manipulation de grandes quantités de données, comme les mesures provenant de capteurs.



## Partie 2 : Création des données


**Objectif** 

Avant de pouvoir analyser des données de capteurs, il faut savoir créer des tableaux de différentes façons : tableaux vides à remplir plus tard, tableaux avec une valeur par défaut, tableaux de temps, ou données aléatoires pour simuler des mesures. 
NumPy propose des fonctions toutes prêtes pour chaque cas.


### 1) Créer un tableau représentant les mesures de température de 10 capteurs, initialement à zéro. 

```python
capteurs_zero = np.zeros(10)
print(capteurs_zero)
```

`np.zeros(n)` crée un tableau de `n` valeurs, toutes égales à 0

### 2) Créer un tableau contenant 10 valeurs égales à 1. 

```python
capteurs_un = np.ones(10)
print(capteurs_un)  
```

`np.ones(n)` fonctionne comme `np.zeros`, mais remplit avec des 1 au lieu de des 0.

### 3) Créer un tableau de 10 valeurs initialisées à 25. 

```python
capteurs_25= np.full(10, 25)
print(capteurs_25)
```

`np.full(n, valeur)` crée un tableau de `n` éléments, tous égaux à `valeur`.

### 4) Créer une matrice identité de taille 5 × 5. 

```python
matrice_identite = np.eye(5)
print(matrice_identite)
```

La **matrice identité** est une matrice carrée avec des 1 sur la diagonale et des 0 partout ailleurs. 

 `np.eye(n)` la génère directement.

### 5) Créer les 24 heures (de 0 à 23) de mesure d’une journée 

```python
heures = np.arange(0, 24, 1)
print(heures)
```

`np.arange(debut, fin)` génère une suite de nombres entiers, comme `range()` en Python classique, mais sous forme de tableau NumPy. Attention : la borne de fin (`24`) n'est **pas incluse**, donc on écrit `np.arange(0, 24)` pour obtenir les valeurs de 0 à 23.

### 6) Créer 10 valeurs régulièrement réparties entre 20 °C et 30 °C  

```python
temperatures= np.linspace(20, 30, 10)
print(temperatures)
```

`np.linspace(debut, fin, nombre_de_valeurs)` crée des valeurs **régulièrement espacées** entre deux bornes. Contrairement à `arange`, ici les deux bornes (20 et 30) **sont incluses**, et on précise directement combien de valeurs on veut au total.

### 7) Créer aléatoirement les températures de 100 mesures, en rendant les résultats reproductibles 

Pour générer des nombres aléatoires, on utilise `np.random`. Mais si on relance le code, on obtiendrait normalement des valeurs différentes à chaque fois !

Pour rendre les résultats **reproductibles** (les mêmes valeurs "aléatoires" à chaque exécution), on fixe une **graine aléatoire** avec `np.random.seed(un_nombre)` juste avant de générer les valeurs.

On utilise ensuite `np.random.uniform(min, max, taille)` pour générer des températures réalistes dans une plage donnée.

```python
np.random.seed(42)
mesures_aleatoires = np.random.uniform(20, 60, 100).round(2)
print(mesures_aleatoires)
```

## Partie 3 – Exploration des tableaux 

**Objectif**

 Avant de manipuler un tableau, il est important de savoir "l'interroger" : combien de dimensions a-t-il, combien de valeurs contient-il, de quel type sont ces valeurs, et quelle est sa forme (nombre de lignes/colonnes). Ce sont des réflexes à avoir systématiquement, surtout quand on récupère des données dont on ne connaît pas encore la structure exacte.

### 1) Créer aléatoirement un tableau représentant les 24 températures d'une journée : 

```python
np.random.seed(42)
temp_jour = np.random.uniform(18, 38, 24).round(2)
print(temp_jour)
```

### 2) Déterminer les propriétés du tableau

Chaque tableau NumPy possède des **attributs** qui décrivent sa structure.On y accède sans parenthèses (ce ne sont pas des fonctions à appeler, mais des informations déjà stockées) :

- `.ndim` : nombre de **dimensions** (1 pour une simple liste de valeurs, 2 pour une matrice, etc.)
- `.size` : nombre total de valeurs dans le tableau
- `.dtype` : type des données stockées (ex: `float64` pour des nombres décimaux)
- `.shape` : la forme du tableau, sous forme de tuple (ex: `(24,)` pour 24 valeurs en 1 dimension)

```python
# a) nombre de dimensions
print("Nombre de dimensions :", temp_jour.ndim)
# b) nombre de valeurs
print("Nombre de valeurs :", temp_jour.size)
# c) type des données
print("Type des données :", temp_jour.dtype)
# d) forme du tableau
print("Type des données :", temp_jour.shape)
```

### Partie 4 – Indexation et slicing 

**Objectif**
 Une fois qu'on a un tableau, on a presque toujours besoin d'en extraire une valeur précise ou une portion. En Python (et donc en NumPy), **le premier élément est à l'index 0**, pas 1. On va utiliser le tableau `temp_jour` créé en Partie 3 (les 24 températures d'une journée, aux heures 0 à 23).

### 1) la première mesure :

On accède à un élément avec des crochets `[index]`. Le premier élément est à l'index `0`.

```python
print(temp_jour[0])  
```

### 2) la dernière mesure :

L'index `-1` désigne toujours le **dernier élément**, `-2` l'avant-dernier, etc. Très pratique pour ne pas avoir à compter le nombre exact d'éléments. 

```python
print(temp_jour[-1]) 
 
```

### 3) la température de 8h :

Comme les heures vont de 0 à 23 et que l'index 0 correspond à l'heure 0, **l'heure et l'index coïncident directement** : l'heure 8h est à l'index `8`.

```python
print(temp_jour[8])
```

### 4) les températures de 8h à 12h :

C'est le **slicing** ("découpage") : `tableau[debut:fin]` extrait une partie du tableau. Attention, la borne de fin est **exclue** : pour aller jusqu'à 12h inclus, il faut écrire `13` comme borne de fin.

```python
print(temp_jour[8:13])
```

### 5) les températures des 6 premières heures :

Quand on omet le `debut` dans un slice, Python part automatiquement du tout début du tableau : `[:6]` équivaut à `[0:6]`.

```python
print(temp_jour[:6])
```

### 6) les températures des heures paires

Le slicing accepte un **3e paramètre**, le pas (`step`) : `tableau[debut:fin:pas]`. Ici on veut les heures 0, 2, 4, 6... donc on part de 0 avec un pas de 2. En laissant `fin` vide, ça va jusqu'au bout du tableau.

```python
print(temp_jour[0::2])
```

## Partie 5 – Filtrage booléen

Jusqu'ici on choisissait des valeurs par leur position (index). Le filtrage booléen permet de choisir des valeurs **selon une condition** — exactement ce dont on a besoin pour détecter des anomalies dans les mesures de capteurs, sans savoir à l'avance à quel index elles se trouvent.

### 1) Afficher les températures supérieures à 30 °C 

```python
print(temp_jour[temp_jour > 30])
```

### 2) Températures « normales » (entre 20°C et 30°C)

On peut combiner plusieurs conditions avec `&` (ET logique). **Important :** en NumPy, chaque condition doit être **entre parenthèses**, et on utilise `&`.

```python
normales = temp_jour[(temp_jour >= 20) & (temp_jour <= 30)]
print(normales)
```

### 3) Supposons qu'une température supérieure à 35 °C soit considérée comme anormale. Déterminer anomalies, un tableau de booléen précisant les anomalies 

Ici on ne filtre pas directement : on **stocke** le résultat de la comparaison dans une variable. `anomalies` sera un tableau de `True`/`False`, de la même taille que `temp_jour`, indiquant pour chaque heure si la température est anormale.

```python
anomalies = temp_jour > 35
print(anomalies)
```

### 4) Déterminer le nombre de températures anormales à l’aide de la variable anomalies  

En Python, `True` est associé à la valeur `1` et `False` à la valeur `0`. Ainsi, lorsque l’on utilise `np.sum()` sur un tableau de valeurs booléennes, la fonction compte automatiquement le nombre de `True`. Cela permet donc de déterminer directement le nombre d’anomalies détectées.


```python
nombre_anomalies = np.sum(anomalies)
print("Nombre d'anomalies :", nombre_anomalies)
```

## Partie6: Manipulation des dimensions 

On suppose qu’on a 5 capteurs. 

Jusqu'ici on travaillait avec un seul capteur (tableau à 1 dimension). Une entreprise a en réalité **plusieurs capteurs** : on a donc besoin de tableaux à 2 dimensions (des **matrices**), où chaque ligne peut représenter un capteur et chaque colonne une heure de la journée (ou l'inverse, selon comment on organise les données).
 ### 1) Créer aléatoirement la matrice représentant les températures mesurées par ces capteurs durant une journée 

On utilise à nouveau `np.random.uniform`, mais cette fois le 3e argument est un **tuple** `(5, 24)` : 5 lignes (les capteurs) et 24 colonnes (les heures). On obtient directement une matrice à 2 dimensions.
e `.round(2)` c'est pour prendre 2 chiffres aprés la virgule

```python
np.random.seed(42)
matrice_5capteurs = np.random.uniform(20, 50, (5, 24)).round(2)
print(matrice_5capteurs)
```

### 2) Transformer la matrice en un tableau unidimensionnel 

`.flatten()` "aplatit" une matrice : toutes les lignes sont mises bout à bout pour former un seul grand tableau à 1 dimension. Ici on obtiendra 5 × 24 = 120 valeurs.

```python
matrice_unidi = matrice_5capteurs.flatten()
print(matrice_unidi)
print("Forme :", matrice_unidi.shape)
```

### 3) Déterminer la transposée de la matrice 

La **transposée** (`.T`) inverse les lignes et les colonnes : une matrice (5 capteurs, 24 heures) devient une matrice (24 heures, 5 capteurs). Utile pour changer de point de vue sur les mêmes données, selon ce qu'on veut analyser.

```python
matrice_5capteurs_transpose = matrice_5capteurs.T
print(matrice_5capteurs_transpose)
```

### 4) Extraire toutes les mesures du capteur 1 

ans notre matrice, chaque **ligne** correspond à un capteur. Le "capteur 1" (au sens humain, le premier) correspond donc à l'**index 0**. `matrice[0]` (ou `matrice[0, :]`) sélectionne toute la première ligne.

```python
capteur_1 = matrice_5capteurs[0, :]
print(capteur_1)
```

### 5) Extraire toutes les mesures du capteur 5 

Le "capteur 5" (5e capteur) correspond à l'index `4` (puisqu'on compte à partir de 0). On peut aussi utiliser l'index `-1` puisque c'est le dernier capteur ici.

```python
capteur_5 = matrice_5capteurs[4, :]
print(capteur_5)
```

### 6) Extraire les mesures de tous les capteurs à 10h 

Cette fois on veut une **colonne** entière (l'heure 10h pour tous les capteurs), pas une ligne. `matrice[:, 10]` signifie "toutes les lignes, colonne d'index 10".

```python
mesures_10h = matrice_5capteurs[:, 10]
print(mesures_10h)
```

### 7) Extraire les mesures des capteurs 2 à 4 entre 8h et 12h

On combine deux slices à la fois : `matrice[lignes, colonnes]`. Les capteurs 2 à 4 correspondent aux index `1` à `3` inclus, donc le slice `1:4` (fin exclue). Les heures 8h à 12h inclus correspondent au slice `8:13`.

```python
extrait = matrice_5capteurs[1:4, 8:13]
print(extrait)
```

##### **À retenir de cette partie :** pour une matrice, l'indexation se fait toujours sous la forme `matrice[lignes, colonnes]`. Le symbole `:` seul signifie "toutes les valeurs de cette dimension". C'est la base pour naviguer dans des données multi-capteurs.

## Partie 8 : Calcul scientifique 

**Pourquoi cette partie ?** NumPy fournit des fonctions statistiques prêtes à l'emploi (au lieu de les recalculer à la main). On les utilise pour résumer et transformer les données de température du 1er bâtiment, avant de les transmettre à un système de Machine Learning — la standardisation et la normalisation sont des étapes très courantes en préparation de données.

### 1) Pour les températures calculer le minimum, le maximum, la somme, la moyenne, la médiane, la variance et l’écart-type 

```python

print("Min :", temp_jour.min())
print("Max :", temp_jour.max())
print("Somme :", temp_jour.sum())
print("Moyenne :", temp_jour.mean())
print("Médiane :", np.median(temp_jour))
print("Variance :", temp_jour.var())
print("Écart-type :", temp_jour.std())
```

### 2) Standardiser le tableau des températures puis vérifier les valeurs de la moyenne et de l’écarttype du tableau standardisé 

La **standardisation** transforme les données pour qu'elles aient une **moyenne de 0** et un **écart-type de 1**. Formule : `(valeur - moyenne) / écart-type`. Très utilisée en Machine Learning, car elle met toutes les variables "à la même échelle".

```python
temperatures_standardisees = (temp_jour - temp_jour.mean()) / temp_jour.std()
print(temperatures_standardisees)
print("Nouvelle moyenne (≈ 0) :", temperatures_standardisees.mean())
print("Nouvel écart-type (≈ 1) :", temperatures_standardisees.std())
```

### 3) Normaliser (Min-Max) le tableau des températures puis vérifier les valeurs du tableau normalisé 


La **normalisation Min-Max** ramène toutes les valeurs entre **0 et 1**. Formule : `(valeur - min) / (max - min)`. La valeur minimale du tableau devient 0, la valeur maximale devient 1, et les autres se répartissent proportionnellement entre les deux.

```python
temperatures_normalisees = (temp_jour - temp_jour.min()) / (temp_jour.max() - temp_jour.min())
print(temperatures_normalisees)
print("Nouveau min (= 0) :", temperatures_normalisees.min())
print("Nouveau max (= 1) :", temperatures_normalisees.max())
```

 ### 4) Convertir en Fahrenheit

Formule de conversion Celsius → Fahrenheit : `F = C × 9/5 + 32`. Grâce à la vectorisation NumPy (vue en Partie 1), on applique directement cette formule à tout le tableau, sans boucle.

```python
temperatures_fahrenheit = temp_jour * 9/5 + 32
print(temperatures_fahrenheit)
```

## Partie 9 : Broadcasting 

Le **broadcasting** est un mécanisme de NumPy qui permet d'effectuer des opérations entre tableaux de tailles différentes, sans avoir à les rendre manuellement de la même taille. C'est extrêmement pratique pour appliquer une correction de calibration à plusieurs capteurs en une seule ligne.

### 1) Matrice `temperatures_3x3` : mesures à 10h, 11h et 12h dans le 2e bâtiment

Le 2e bâtiment a 3 capteurs, et on s'intéresse à 3 heures précises (10h, 11h, 12h). On organise la matrice avec **une ligne par heure** et **une colonne par capteur** : forme `(3 heures, 3 capteurs)`. C'est ce choix d'organisation (heures en lignes, capteurs en colonnes) qui va permettre au broadcasting de bien fonctionner à l'étape suivante.

```python
np.random.seed(3)
temperatures_3x3 = np.random.uniform(15, 35, (3, 3))  # 3 heures (lignes) x 3 capteurs (colonnes)
print(temperatures_3x3)
print("Forme :", temperatures_3x3.shape)
```

### 2) Appliquer la correction de calibration

`correction` est un tableau de 3 valeurs (une par capteur), alors que `temperatures_3x3` est une matrice `(3, 3)`. Normalement, additionner deux tableaux de tailles différentes devrait poser problème...

...mais grâce au **broadcasting**, NumPy comprend automatiquement que `correction` doit être appliqué **à chaque ligne** (donc à chaque heure), colonne par colonne — c'est-à-dire capteur par capteur. C'est comme si `correction` était "répété" 3 fois, une fois par ligne (par heure), sans qu'on ait besoin de l'écrire explicitement. Résultat : le capteur 1 (1ère colonne) reçoit toujours `+1`, le capteur 2 toujours `-0.5`, le capteur 3 toujours `+0.2`, et ce pour les 3 heures.

```python
correction = np.array([1, -0.5, 0.2])
mesures_corrigees = temperatures_3x3 + correction
print(mesures_corrigees)
```
